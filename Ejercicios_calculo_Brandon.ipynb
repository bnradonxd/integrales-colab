```python
def f(x):
  return (4 * x**2 - 5 * x + 2)**3

def f_prima(x):
    return 3 * (4 * x**2 - 5 * x + 2)**2 * (8 * x * - 5)

N = 100
x = np.linspace( -1 , 3 , N)

y = f(x)
y_prima = f_prima(x)

plt.plot(x, y, label="$f(x) = (4x^2 - 5x + 2)^3$")
plt.plot(x, y_prima, label="$f'(x) (derivada)$", linestyle="--")
plt.grid()

plt.xlabel("Eje X")
plt.ylabel("Eje Y")

plt.legend()

plt.title("Gráfica de la función $f(x)$ y $f'(x)$ (derivada)")
plt.show()
```

# Integración

### Librerias

```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, diff, integrate, lambdify, sqrt, sin, cos
import pandas as pd
```

### Ejercicio 1

```python
x = symbols('x')

fx = x**2 - 9*x

derivada_fx = diff(fx, x)
integral_indef_fx = integrate(fx, x)

f_num = lambdify(x, fx, 'numpy')
derivada_num = lambdify(x, derivada_fx, 'numpy')
integral_indef_num = lambdify(x, integral_indef_fx, 'numpy')

x_val = np.linspace(1, 4, 100)

y_val = f_num(x_val)
dy_val = derivada_num(x_val)
iy_val = integral_indef_num(x_val)

integral_def = integrate(fx, (x, 1, 4))

plt.figure(figsize=(10, 6))
plt.plot(x_val, y_val, label=r"$f(x) = x^2 - 9x$", color="blue")
plt.plot(x_val, dy_val, label=r"$f'(x)$ (Derivada de $f(x)$)", linestyle="--", color="orange")
plt.plot(x_val, iy_val, label=r"$\int f(x) \, dx$ (Integral indefinida de $f(x)$)", linestyle=":", color="red")

plt.fill_between(
    x_val,
    y_val,
    where=(x_val >= 1) & (x_val <= 4),
    color="lightblue",
    alpha=0.5,
    label=f"Área bajo la curva (Integral definida = {integral_def:.2f})",
)

plt.title("Gráfica de $f(x)$, su derivada y su integral indefinida")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.legend()
plt.grid(True)



plt.show()
```
